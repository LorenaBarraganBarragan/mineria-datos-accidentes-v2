5. Integración de los tres datasets

5.1 Preparación de las fechas

```python
accidentes_limpios = pd.read_csv(
    "../dataset/Accidentalidad_en_Barranquilla_limpio.csv"
)

precipitacion_limpia = pd.read_csv(
    "../dataset/precipitacion_barranquilla_limpio.csv"
)

comparendos_limpios = pd.read_csv(
    "../dataset/Comparendos_Barranquilla_limpio.csv"
)
```

5.1.1 Preparación de fecha en accidentalidad

```python
# Limpiar el formato de la hora
hora_limpia = (
    accidentes_limpios["HORA_ACCIDENTE"]
    .str.replace(":am", " AM", regex=False)
    .str.replace(":pm", " PM", regex=False)
)

# Combinar fecha y hora
accidentes_limpios["FECHA_HORA"] = pd.to_datetime(
    accidentes_limpios["FECHA_ACCIDENTE"].str[:10] + " " + hora_limpia,
    format="%Y-%m-%d %I:%M:%S %p",
    errors="coerce"
)
```

```python
accidentes_limpios[
    ["FECHA_ACCIDENTE", "HORA_ACCIDENTE", "FECHA_HORA"]
].head()
```

5.1.2 Preparación de fecha en precipitación

```python
precipitacion_limpia["FECHA_HORA"] = pd.to_datetime(
    precipitacion_limpia["fechaobservacion"],
    errors="coerce"
)
```

```python
precipitacion_limpia[
    ["fechaobservacion", "FECHA_HORA"]
].head()
```

5.1.3 Preparación de fecha en comparendos

```python
comparendos_limpios["FECHA_HORA"] = pd.to_datetime(
    comparendos_limpios["fecha_comparendo"],
    errors="coerce"
)
```

```python
comparendos_limpios[
    ["fecha_comparendo", "FECHA_HORA"]
].head()
```

5.2 Relación entre los datasets

5.2.1 Período común de los tres datasets

```python
print("ACCIDENTES")
print(
    accidentes_limpios["FECHA_HORA"].min(),
    "a",
    accidentes_limpios["FECHA_HORA"].max()
)

print("\nPRECIPITACIÓN")
print(
    precipitacion_limpia["FECHA_HORA"].min(),
    "a",
    precipitacion_limpia["FECHA_HORA"].max()
)

print("\nCOMPARENDOS")
print(
    comparendos_limpios["FECHA_HORA"].min(),
    "a",
    comparendos_limpios["FECHA_HORA"].max()
)
```

5.2.2 Filtrar los tres datasets al período común

```python
fecha_inicio = "2018-01-01"
fecha_fin = "2019-08-27 23:59:59"

accidentes_periodo = accidentes_limpios[
    (accidentes_limpios["FECHA_HORA"] >= fecha_inicio) &
    (accidentes_limpios["FECHA_HORA"] <= fecha_fin)
].copy()

precipitacion_periodo = precipitacion_limpia[
    (precipitacion_limpia["FECHA_HORA"] >= fecha_inicio) &
    (precipitacion_limpia["FECHA_HORA"] <= fecha_fin)
].copy()

comparendos_periodo = comparendos_limpios[
    (comparendos_limpios["FECHA_HORA"] >= fecha_inicio) &
    (comparendos_limpios["FECHA_HORA"] <= fecha_fin)
].copy()
```

```python
print("Accidentes:", len(accidentes_periodo))
print("Precipitación:", len(precipitacion_periodo))
print("Comparendos:", len(comparendos_periodo))
```

5.2.3 Agrupación de los datos por fecha

```python
accidentes_periodo["FECHA"] = accidentes_periodo["FECHA_HORA"].dt.date

precipitacion_periodo["FECHA"] = precipitacion_periodo["FECHA_HORA"].dt.date

comparendos_periodo["FECHA"] = comparendos_periodo["FECHA_HORA"].dt.date
```

```python
print("Fechas de accidentes:", accidentes_periodo["FECHA"].nunique())
print("Fechas de precipitación:", precipitacion_periodo["FECHA"].nunique())
print("Fechas de comparendos:", comparendos_periodo["FECHA"].nunique())
```

```python
import pandas as pd

# Cargar datasets
accidentes = pd.read_csv("../dataset/Accidentalidad_en_Barranquilla_20260830.csv")
precipitacion = pd.read_csv("../dataset/precipitacion_barranquilla.csv")
comparendos = pd.read_csv("../dataset/comparendos_Barranquilla.csv")
```

```python
print("ACCIDENTES")
print(accidentes.columns.tolist())

print("\nPRECIPITACIÓN")
print(precipitacion.columns.tolist())

print("\nCOMPARENDOS")
print(comparendos.columns.tolist())
```

```python
print("Accidentes:")
display(accidentes.head())

print("Precipitación:")
display(precipitacion.head())

print("Comparendos:")
display(comparendos.head())
```

```python
# Convertir las fechas al mismo formato: solo fecha

accidentes["FECHA"] = pd.to_datetime(
    accidentes["FECHA_ACCIDENTE"],
    errors="coerce"
).dt.date

precipitacion["FECHA"] = pd.to_datetime(
    precipitacion["fechaobservacion"],
    errors="coerce"
).dt.date

comparendos["FECHA"] = pd.to_datetime(
    comparendos["fecha_comparendo"],
    errors="coerce",
    format="mixed"
).dt.date
```

```python
accidentes_dia = accidentes.groupby("FECHA").agg(
    ACCIDENTES=("CANTIDAD_ACCIDENTES", "sum"),
    GRAVEDAD_ACCIDENTE=("GRAVEDAD_ACCIDENTE", lambda x: x.mode()[0] if not x.mode().empty else "Sin dato"),
    CLASE_ACCIDENTE=("CLASE_ACCIDENTE", lambda x: x.mode()[0] if not x.mode().empty else "Sin dato")
).reset_index()

display(accidentes_dia.head())
```

```python
precipitacion_dia = precipitacion.groupby("FECHA").agg(
    PRECIPITACION_MM=("valorobservado", "sum")
).reset_index()

precipitacion_dia["LLUVIA"] = precipitacion_dia["PRECIPITACION_MM"].apply(
    lambda x: "Sí" if x > 0 else "No"
)

display(precipitacion_dia.head())
```

```python
comparendos_dia = comparendos.groupby("FECHA").agg(
    COMPARENDOS=("CANTIDAD_INFRACCIONES", "sum"),
    COD_INFRACCION=("COD_INFRACCION", lambda x: x.mode()[0] if not x.mode().empty else "Sin dato"),
    TIPO_INFRACCION=("TIPO_INFRACCION", lambda x: x.mode()[0] if not x.mode().empty else "Sin dato"),
    CLASE_VEHICULO_INFRACTOR=("CLASE_VEHICULO_INFRACTOR", lambda x: x.mode()[0] if not x.mode().empty else "Sin dato"),
    SERVICIO_VEHICULO_INFRACTOR=("SERVICIO_VEHICULO_INFRACTOR", lambda x: x.mode()[0] if not x.mode().empty else "Sin dato")
).reset_index()

display(comparendos_dia.head())
```

```python
dataset_integrado = accidentes_dia.merge(
    precipitacion_dia,
    on="FECHA",
    how="outer"
).merge(
    comparendos_dia,
    on="FECHA",
    how="outer"
)

dataset_integrado = dataset_integrado.sort_values("FECHA").reset_index(drop=True)

display(dataset_integrado.head(10))
```

```python
print("Filas y columnas:", dataset_integrado.shape)

print("\nColumnas:")
for columna in dataset_integrado.columns:
    print("-", columna)
```

```python
columnas_categoricas = [
    "GRAVEDAD_ACCIDENTE",
    "CLASE_ACCIDENTE",
    "LLUVIA",
    "COD_INFRACCION",
    "TIPO_INFRACCION",
    "CLASE_VEHICULO_INFRACTOR",
    "SERVICIO_VEHICULO_INFRACTOR"
]

for columna in columnas_categoricas:
    df_transformado[columna] = df_transformado[columna].fillna("Sin dato")
```

Uso categoria ordinal para GRAVEDAD_ACCIDENTE

```python
mapa_gravedad = {
    "Solo daños": 0,
    "Con heridos": 1,
    "Con muertos": 2,
    "Sin dato": -1
}

df_transformado["GRAVEDAD_ACCIDENTE"] = (
    df_transformado["GRAVEDAD_ACCIDENTE"].map(mapa_gravedad)
)
```

Para las demas uso One-Hot: CLASE_ACCIDENTE, LLUVIA, COD_INFRACCION, TIPO_INFRACCION, CLASE_VEHICULO_INFRACTOR y SERVICIO_VEHICULO_INFRACTOR

```python
columnas_one_hot = [
    "CLASE_ACCIDENTE",
    "LLUVIA",
    "COD_INFRACCION",
    "TIPO_INFRACCION",
    "CLASE_VEHICULO_INFRACTOR",
    "SERVICIO_VEHICULO_INFRACTOR"
]

df_transformado = pd.get_dummies(
    df_transformado,
    columns=columnas_one_hot,
    dtype=int
)
```

Probar los 3 escaladores 

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler

# Variables numéricas que vamos a escalar
variables_numericas = [
    "ACCIDENTES",
    "PRECIPITACION_MM",
    "COMPARENDOS"
]

# Crear copias de los datos
datos_numericos = df_transformado[variables_numericas].copy()

# 1. MinMaxScaler
minmax = MinMaxScaler()
datos_minmax = pd.DataFrame(
    minmax.fit_transform(datos_numericos),
    columns=variables_numericas
)

# 2. StandardScaler
standard = StandardScaler()
datos_standard = pd.DataFrame(
    standard.fit_transform(datos_numericos),
    columns=variables_numericas
)

# 3. RobustScaler
robust = RobustScaler()
datos_robust = pd.DataFrame(
    robust.fit_transform(datos_numericos),
    columns=variables_numericas
)

print("MINMAX")
display(datos_minmax.describe())

print("STANDARD")
display(datos_standard.describe())

print("ROBUST")
display(datos_robust.describe())
```

Voy a usar RobustScaler
¿Por que?
Se seleccionó RobustScaler para las variables numéricas ACCIDENTES, PRECIPITACION_MM y COMPARENDOS, debido a la presencia de valores extremos y distribuciones con alta concentración de valores en determinados rangos. RobustScaler utiliza la mediana y el rango intercuartílico, por lo que es menos sensible a valores extremos.

Revisar valores nulos de COMPARENDOS 

```python
print("Nulos en COMPARENDOS:", df_transformado["COMPARENDOS"].isna().sum())
print("Mediana:", df_transformado["COMPARENDOS"].median())
```

```python
# Imputar valores faltantes de COMPARENDOS con la mediana
mediana_comparendos = df_transformado["COMPARENDOS"].median()

df_transformado["COMPARENDOS"] = (
    df_transformado["COMPARENDOS"].fillna(mediana_comparendos)
)

print("Mediana utilizada:", mediana_comparendos)
print("Nulos restantes:", df_transformado["COMPARENDOS"].isna().sum())
```

Aplicar RobustScaler

```python
from sklearn.preprocessing import RobustScaler

# Variables numéricas que vamos a escalar
variables_numericas = [
    "ACCIDENTES",
    "PRECIPITACION_MM",
    "COMPARENDOS"
]

# Crear el escalador
scaler = RobustScaler()

# Aplicar el escalamiento
df_transformado[variables_numericas] = scaler.fit_transform(
    df_transformado[variables_numericas]
)

print("Escalamiento realizado correctamente.")
display(df_transformado[variables_numericas].describe())
```

```python
print(df_transformado.columns.tolist())
```

Crear las variables para PCA

```python
# Seleccionar variables numéricas para PCA
variables_pca = df_transformado.select_dtypes(
    include="number"
).columns.tolist()

# Excluir el indicador de datos faltantes
variables_pca.remove("PRECIPITACION_FALTANTE")

print("Variables que entrarán al PCA:")
print(variables_pca)
print("\nCantidad de variables:", len(variables_pca))
```

Calcular PCA completo

```python
from sklearn.decomposition import PCA

# Datos que entrarán al PCA
datos_pca = df_transformado[variables_pca]

# Crear PCA sin limitar componentes
pca = PCA()

# Ajustar PCA
pca.fit(datos_pca)

# Varianza explicada por cada componente
varianza_explicada = pca.explained_variance_ratio_

print("Varianza explicada por componente:")
for i, varianza in enumerate(varianza_explicada, start=1):
    print(f"PC{i}: {varianza:.4f} ({varianza * 100:.2f}%)")
```

```python
# Calcular la varianza acumulada
varianza_acumulada = pca.explained_variance_ratio_.cumsum()

for i, varianza in enumerate(varianza_acumulada, start=1):
    print(f"PC{i}: {varianza:.4f} ({varianza * 100:.2f}%)")
```

Pasamos de 24 variables → 5 componentes, conservando aproximadamente 90,83 % de la información.

```python
from sklearn.decomposition import PCA

# Reducir a 5 componentes principales
pca_reducido = PCA(n_components=5)

datos_reducidos = pca_reducido.fit_transform(
    df_transformado[variables_pca]
)

# Crear DataFrame reducido
df_pca = pd.DataFrame(
    datos_reducidos,
    columns=["PC1", "PC2", "PC3", "PC4", "PC5"],
    index=df_transformado.index
)

print("Dimensiones del dataset reducido:", df_pca.shape)

display(df_pca.head())
```

```python
# Crear dataset final reducido
df_reducido = pd.concat(
    [
        df_transformado[["FECHA", "PRECIPITACION_FALTANTE"]],
        df_pca
    ],
    axis=1
)

print("Dimensiones del dataset final:", df_reducido.shape)

display(df_reducido.head())
```

Guarda el dataset reducido

```python
# Guardar dataset final reducido
ruta_salida = "../dataset/dataset_reducido.csv"

df_reducido.to_csv(ruta_salida, index=False)

print(f"Dataset guardado correctamente en: {ruta_salida}")
```

Gráfica de varianza acumulada

```python
import matplotlib.pyplot as plt

varianza_acumulada = pca.explained_variance_ratio_.cumsum()

plt.figure(figsize=(8, 5))
plt.plot(
    range(1, len(varianza_acumulada) + 1),
    varianza_acumulada * 100,
    marker="o"
)

plt.axhline(y=90, linestyle="--")
plt.axvline(x=5, linestyle="--")

plt.xlabel("Número de componentes")
plt.ylabel("Varianza acumulada (%)")
plt.title("Varianza acumulada del PCA")
plt.grid(True)

plt.show()
```

Gráfica del patrón comportamental

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.scatter(
    df_pca["PC1"],
    df_pca["PC2"],
    alpha=0.5
)

plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("Patrón comportamental de los datos: PC1 vs PC2")
plt.grid(True)

plt.show()
```

Patrón comportamental: Después de aplicar PCA, los datos presentan una distribución dispersa entre los componentes PC1 y PC2. Los registros se encuentran distribuidos en los cuatro cuadrantes, sin una concentración dominante en uno solo. Esto indica que el comportamiento de los registros está influenciado por diferentes combinaciones de las variables originales.
